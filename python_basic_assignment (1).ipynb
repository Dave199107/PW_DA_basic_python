# **Python Basic questions**

**1. What is Python, and why is it popular?**

Python is an easy-to-learn programming language used for web development, data science, automation, and more. It’s popular because of its simple syntax, readability, and huge community support.

**2. What is an interpreter in Python?**

An interpreter (is a transalator) reads and executes Python code line by line (unlike compiled languages that convert the whole code at once).

**3. What are pre-defined keywords in Python?**

Keywords are reserved words with special meanings (e.g., if, else, for, while). They cannot be used as variable names.

**4. Can keywords be used as variable names?**

No, keywords like if, else, for cannot be used as variable names.

**5. What is mutability in Python?**

Mutability means whether an object can be changed after creation.

Mutable = Can change (e.g., lists, dictionaries)

Immutable = Cannot change (e.g., tuples, strings, numbers)

**6. Why are lists mutable, but tuples are immutable?**

Lists are mutable (can add/remove items) → Useful for dynamic data.

Tuples are immutable (fixed size) → Faster and safer for constant data.

**7. What is the difference between == and is operators?**

== checks value equality.

**is** checks memory location (same object).

**8. What are logical operators in Python?**

They combine conditions:

and → Both must be True.

or → At least one True.

not → Reverses the condition.

**9. What is type casting in Python?**

Converting one data type to another (e.g., int("5") converts string to integer).

**10. Difference between implicit and explicit type casting?**

Implicit: Python does it automatically (e.g., 5 + 2.0 → 7.0).

Explicit: Programmer does it manually (e.g., float(5) → 5.0).

**11. What is the purpose of conditional statements?**

They make decisions in code (e.g., if-else runs code only if a condition is met).

**12. How does elif work?**

elif (else-if) checks multiple conditions after if:

python
if x > 10:  
    print("Big")  
elif x > 5:  
    print("Medium")  
else:  
    print("Small")  
    
**13. Difference between for and while loops?**

**for loop:** Runs for a fixed number of times (e.g., over a list).

**while loop:** Runs until a condition is False.

**14. When to use while instead of for?**

Use **while** when you don’t know how many times the loop should run (e.g., asking for user input until it’s correct).

6. Write a function to demonstrate the behavior of mutable and immutable arguments.

```python
def modify_data(a_list, a_string):
    a_list.append(4)  # Mutable (changes original list)
    a_string += " World"  # Immutable (new string created)
    print("Inside function:", a_list, a_string)

my_list = [1, 2, 3]
my_string = "Hello"
modify_data(my_list, my_string)
print("Outside function:", my_list, my_string)
```

13. Write a Python program to find the sum of all even numbers between 1 and 50.

```python
sum_even = 0
for num in range(2, 51, 2):  # Start at 2, step by 2
    sum_even += num
print("Sum of even numbers:", sum_even)
```

14. Write a program to reverse a string using a while loop.

```python
text = input("Enter a string: ")
reversed_text = ""
index = len(text) - 1
while index >= 0:
    reversed_text += text[index]
    index -= 1
print("Reversed:", reversed_text)
```

15. Write a Python program to calculate the factorial of a number provided by the user using a while loop.

```python
num = int(input("Enter a number: "))
factorial = 1
i = 1
while i <= num:
    factorial *= i
    i += 1
print(f"Factorial of {num} is {factorial}")
```
